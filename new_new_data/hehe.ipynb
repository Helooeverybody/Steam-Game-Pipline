```python
from collections import defaultdict
hehe = defaultdict(int)
import json
import os
i=0
with open("top_10_reviews_raw/578080.jsonl", "r") as f:
    while True:
        print(i)
        i+=1
        json_str = f.readline()
        if f.tell() == os.fstat(f.fileno()).st_size:
            break
        result = json.loads(json_str)
        lang = result["language"]
        hehe[lang] += 1

print(hehe)
```
