```python
import json
```

```python
with open('steam_apps_dataset_raw.jsonl', 'r', encoding='utf-8') as f:
    for line in f:
        hehe = line
        break
    
print(json.loads(hehe))
```

```python
k = json.loads(hehe)
# list all keys of k, which k is nested dict
for key in k.keys():
    print(key)
    print(f"{key}: {type(k[key])}")
    if isinstance(k[key], dict):
        for subkey in k[key].keys():
            print(f"  {subkey}: {type(k[key][subkey])}")
```
